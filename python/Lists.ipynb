# Lists in Python

A list in Python is a built-in data structure used to store collections of data. Lists are ordered, mutable (changeable), and allow duplicate values.

## When to use a list
You should use a list when:
- You need to maintain the order of items.
- You have a collection of items that might need to be modified (added, removed, or changed) frequently.
- You want to store duplicates of data.
- You need to access elements by their integer index.

## Inbuilt Methods of Lists
Python provides several built-in methods to manipulate lists:
- `append(item)`: Adds an item to the end of the list.
- `extend(iterable)`: Adds all elements of an iterable (like another list) to the end.
- `insert(index, item)`: Inserts an item at a specific index.
- `remove(item)`: Removes the first occurrence of an item.
- `pop(index)`: Removes and returns the item at the given index (or the last item if no index is specified).
- `sort()`: Sorts the list in ascending order.
- `reverse()`: Reverses the order of the list in place.
- `index(item)`: Returns the index of the first matched item.
- `count(item)`: Returns the number of times an item appears in the list.

## How to Iterate a List
You can iterate through a list using a `for` loop. This can be done either directly on the items, or by using indices with `range()` and `len()`.

```python
# --- Examples of List Methods ---
my_list = [10, 20, 30]

# append()
my_list.append(40)
print("After append:", my_list)

# insert()
my_list.insert(1, 15)
print("After insert:", my_list)

# remove()
my_list.remove(20)
print("After remove:", my_list)

# --- Iterating a List ---
print("\nIterating directly:")
for item in my_list:
    print(item)

print("\nIterating using index:")
for i in range(len(my_list)):
    print(f"Index {i}: {my_list[i]}")
```

# List Comprehension

List comprehension offers a shorter and more compact syntax when you want to create a new list based on the values of an existing iterable (like a list, tuple, or string).

## Why to use it
- **Readability**: It is often more concise and easier to read than a standard `for` loop.
- **Performance**: It is generally faster and more computationally efficient than traditional loops because it is optimized in C.
- **Elegance**: It condenses multiple lines of code (creating an empty list, looping, and appending) into a single, clean line.

## Basic Syntax
`new_list = [expression for item in iterable]`

## Syntax with Conditionals
`new_list = [expression for item in iterable if condition]`
You can add an `if` condition to the end to filter which items are included or processed.

## Syntax with Function Calls
`new_list = [function(item) for item in iterable]`
The expression part can be any valid expression, including calling a function on each item.

```python
# 1. Basic List Comprehension
numbers = [1, 2, 3, 4, 5]

# Create a new list containing the squares of the numbers
squares = [n**2 for n in numbers]

print("Original numbers:", numbers)
print("Squares (Basic):", squares)
```

```python
# 2. List Comprehension with Conditionals

# Filter out only the even numbers from the original list
evens = [n for n in numbers if n % 2 == 0]

print("Even numbers (Conditionals):", evens)
```

```python
# 3. List Comprehension with Function Calls

def double_value(x):
    return x * 2

# Apply the double_value function to each item in the list
doubled = [double_value(n) for n in numbers]

print("Doubled using function:", doubled)
```

# List Slicing

List slicing allows you to extract a portion (a sub-list) from an existing list. It returns a new list containing the requested elements without modifying the original.

**Basic Syntax**: `my_list[start:stop:step]`
- `start`: The starting index (inclusive). Defaults to `0` if omitted.
- `stop`: The ending index (exclusive). Defaults to the end of the list if omitted.
- `step`: The step or stride (how many elements to jump). Defaults to `1` if omitted.

## Common Slicing Patterns
- `my_list[1:4]`: Elements from index 1 up to (but not including) index 4.
- `my_list[:1]`: Just the first element, but returned as a list.
- `my_list[:3]`: The first 3 elements (from start up to index 3 exclusive).
- `my_list[2:]`: From index 2 to the end of the list.
- `my_list[:]`: A shallow copy of the entire list.
- `my_list[::2]`: Every second element (step of 2) across the whole list.
- `my_list[::-1]`: The entire list in reverse order.

```python
# --- List Slicing Examples ---
letters = ['a', 'b', 'c', 'd', 'e', 'f', 'g']

print("Original:", letters)
print("1 to 4 (letters[1:4]):", letters[1:4])       # ['b', 'c', 'd']
print("Up to index 1 (letters[:1]):", letters[:1])  # ['a'] (different from letters[0] which is a string)
print("Up to index 3 (letters[:3]):", letters[:3])  # ['a', 'b', 'c']
print("Index 4 to end (letters[4:]):", letters[4:]) # ['e', 'f', 'g']
print("Every 2nd item (letters[::2]):", letters[::2]) # ['a', 'c', 'e', 'g']
print("Reverse list (letters[::-1]):", letters[::-1]) # ['g', 'f', 'e', 'd', 'c', 'b', 'a']
```

# Common Errors to Avoid with Lists

1. **`IndexError`**: Trying to access an index that doesn't exist (e.g., `my_list[10]` in a list of 3 items). Remember that Python lists are 0-indexed.
2. **Modifying a List While Iterating Over It**: Removing or adding items to a list while iterating with a `for` loop can cause unexpected behavior (skipping items). **Solution**: Iterate over a copy of the list (e.g., `for item in my_list[:]`).
3. **Unexpected Behavior with Shared References**: Using `list2 = list1` does not copy the list; it just creates another reference to the exact same list in memory. Modifying `list2` will also modify `list1`.

```python
# --- Common Errors Examples ---
my_nums = [10, 20, 30]

# 1. IndexError (Uncommenting the line below will crash the program)
# print(my_nums[5]) 

# 2. Modifying while iterating (The wrong way vs the right way)
# The RIGHT way:
for num in my_nums[:]:  # Iterate over a slice (a shallow copy)
    if num == 20:
        my_nums.remove(num)
print("After safe removal:", my_nums)

# 3. Shared References
list_a = [1, 2, 3]
list_b = list_a  # This is NOT a copy! It's a shared reference.
list_b.append(4)
print("list_a is also modified! list_a:", list_a)
```

# Other Important List Concepts

## 1. Nested Lists (2D Lists / Matrices)
A list can contain other lists as its elements. This is useful for representing matrices or grids.

## 2. Copying Lists Properly
To avoid the shared reference issue mentioned above, you must explicitly copy the list:
- **Shallow Copy**: Good for flat lists. Can be done using `list2 = list1.copy()`, `list2 = list1[:]`, or `list2 = list(list1)`.
- **Deep Copy**: Crucial when you have nested lists. A shallow copy only copies the outer list, leaving the inner lists as shared references. Use `copy.deepcopy()` to recursively copy all nested elements.

```python
# --- Nested Lists ---
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print("Row 0, Col 1 of matrix:", matrix[0][1])  # Output: 2

# --- Copying Lists Properly ---
import copy

original = [[1, 2], [3, 4]]

# Shallow Copy
shallow = original.copy()
shallow[0].append(99)
print("\nOriginal after modifying shallow copy:", original) # Modifies original's nested list!

# Deep Copy
deep = copy.deepcopy(original)
deep[1].append(100)
print("Original after modifying deep copy:", original) # Stays the same!
```
