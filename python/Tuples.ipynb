# Python Tuples

A tuple in Python is a collection which is ordered and unchangeable (immutable). 
Tuples are written with round brackets `()`.

## Key Characteristics:
- **Ordered:** The items have a defined order, and that order will not change.
- **Unchangeable (Immutable):** We cannot change, add, or remove items after the tuple has been created.
- **Allow Duplicates:** Since tuples are indexed, they can have items with the same value.
- **Heterogeneous:** Tuples can contain items of different data types (e.g., integers, strings, booleans, etc.).

## Creating a Tuple:
```python
# Empty tuple
my_tuple = ()

# Tuple with values
my_tuple = ("apple", "banana", "cherry")

# Tuple with a single item (comma is mandatory)
single_item_tuple = ("apple",) 
# Note: ("apple") without a comma is evaluated as a string.
```

## Accessing Tuple Items:
You can access tuple items by referring to the index number, inside square brackets.
- Positive indexing starts from `0`.
- Negative indexing starts from `-1` (last item).
- You can also specify a range of indexes (slicing) like `my_tuple[1:3]`.

````python
# --- Converting Tuple to List and List to Tuple ---

# 1. Converting a Tuple to a List
# Since tuples are immutable, a common workaround to modify a tuple is to 
# convert it into a list, change the list, and convert it back to a tuple.
my_tuple = ("apple", "banana", "cherry")
print(f"Original Tuple: {my_tuple} (Type: {type(my_tuple)})")

# Use the list() constructor
my_list = list(my_tuple)
print(f"Converted List: {my_list} (Type: {type(my_list)})")

# Now we can modify the list
my_list.append("orange")
my_list[0] = "kiwi"

# 2. Converting a List back to a Tuple
# Use the tuple() constructor
modified_tuple = tuple(my_list)
print(f"Modified Tuple: {modified_tuple} (Type: {type(modified_tuple)})")
````

# Unpacking Tuples and Nested Tuples

## 1. Unpacking a Tuple
When we create a tuple, we normally assign values to it. This is called "packing" a tuple.
In Python, we are also allowed to extract the values back into variables. This is called "unpacking".

````python
# Packing a tuple
fruits = ("apple", "banana", "cherry")

# Unpacking a tuple (number of variables must match number of items)
(green, yellow, red) = fruits
print(f"green: {green}")
print(f"yellow: {yellow}")
print(f"red: {red}")
````

## 2. Unpacking using Asterisk `*`
If the number of variables is less than the number of values, you can add an `*` to the variable name and the values will be assigned to the variable as a list.

````python
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")

# Unpacking using *
(green, yellow, *red) = fruits
print(f"green: {green}")
print(f"yellow: {yellow}")
print(f"red: {red}") # red gets the rest of the items as a list

print("\n--- Unpacking with * in the middle ---")
# The * can be used on any variable
(green, *tropic, red) = fruits
print(f"green: {green}")
print(f"tropic: {tropic}")
print(f"red: {red}")
````

## 3. Nested Tuples
A tuple can contain another tuple as an element. This is called a nested tuple.

````python
# Creating a nested tuple
nested_tuple = ("mouse", [8, 4, 6], (1, 2, 3))

print(f"Nested Tuple: {nested_tuple}")

# Accessing elements in a nested tuple
print(f"Accessing list element (nested_tuple[1][1]): {nested_tuple[1][1]}") # Output: 4
print(f"Accessing tuple element (nested_tuple[2][2]): {nested_tuple[2][2]}") # Output: 3
````

# Common Mistakes to Avoid with Tuples

1. **Forgetting the comma when creating a single-element tuple:**
   - **Wrong:** `my_tuple = ("apple")` (This creates a string)
   - **Right:** `my_tuple = ("apple",)` (This creates a tuple)

2. **Trying to modify a tuple directly (Immutability):**
   - **Wrong:** `my_tuple[0] = "orange"` -> Raises a `TypeError: 'tuple' object does not support item assignment`
   - **Solution:** Convert to a list, modify the list, and convert back to a tuple.

3. **Mismatched variables during unpacking:**
   - **Wrong:** `(a, b) = (1, 2, 3)` -> Raises a `ValueError: too many values to unpack`
   - **Solution:** Ensure the number of variables matches the items, or use the `*` operator.

4. **Assuming tuples are entirely immutable:**
   - While the tuple itself cannot be modified (cannot add/remove elements or change which object an index points to), mutable objects *inside* the tuple CAN be modified.
   - For example: `my_tuple = (1, 2, [3, 4])`. You cannot reassign `my_tuple[2]`, but you CAN append to the list: `my_tuple[2].append(5)`.

# Tuple Methods

Python has two built-in methods that you can use on tuples:

1. `count()`: Returns the number of times a specified value occurs in a tuple.
2. `index()`: Searches the tuple for a specified value and returns the position of where it was found.

````python
my_tuple = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5, 7)

# 1. count() method
count_sevens = my_tuple.count(7)
print(f"The number 7 appears {count_sevens} times in the tuple.")

# 2. index() method
first_eight_index = my_tuple.index(8)
print(f"The first occurrence of the number 8 is at index: {first_eight_index}")

# Note: index() raises a ValueError if the value is not found in the tuple.
````
