# Python Dictionaries

## 1. Introduction to Dictionaries
A dictionary in Python is a collection of key-value pairs. Each key is connected to a value, and you can use a key to access the value associated with it.

**Key Characteristics:**
- **Ordered:** As of Python 3.7, dictionaries maintain insertion order.
- **Changeable (Mutable):** You can change, add, or remove items after the dictionary is created.
- **Unique Keys:** Dictionaries cannot have two items with the same key. Keys must be of an immutable data type (like strings, numbers, or tuples).

## 2. Creating Dictionaries
You can create a dictionary by enclosing a comma-separated list of key-value pairs in curly braces `{}`, or by using the `dict()` constructor.

```python
# Creating an empty dictionary
empty_dict = {}

# Creating a dictionary with curly braces
student = {
    "name": "Alice",
    "age": 22,
    "major": "Computer Science"
}
print("Student Dict:", student)

# Creating a dictionary using the dict() constructor
car = dict(brand="Ford", model="Mustang", year=1964)
print("Car Dict:", car)
```

## 3. Accessing Dictionary Elements
You can access the items of a dictionary by referring to its key name, inside square brackets. Alternatively, you can use the `.get()` method, which is safer as it won't raise an error if the key doesn't exist.

```python
person = {"name": "Bob", "age": 30, "city": "New York"}

# Accessing using square brackets
print("Name:", person["name"])

# Accessing using the get() method
print("Age:", person.get("age"))

# get() allows setting a default value if the key is not found
print("Country:", person.get("country", "Not Specified"))

# Getting all keys, values, and items
print("Keys:", person.keys())
print("Values:", person.values())
print("Items (Key-Value pairs):", person.items())
```

## 4. Modifying Dictionary Elements
You can add new elements, update existing values, and remove elements using various built-in approaches.

```python
inventory = {"apples": 10, "bananas": 5}

# Updating an existing element
inventory["apples"] = 15

# Adding a new element
inventory["oranges"] = 8
print("Updated Inventory:", inventory)

# Removing an element using pop() (returns the value)
removed_item = inventory.pop("bananas")
print(f"Removed bananas ({removed_item}). Current Inventory:", inventory)

# Removing the last inserted item using popitem()
inventory.popitem()
print("After popitem():", inventory)

# Removing an element using the del keyword
del inventory["apples"]
print("After del:", inventory)
```

## 5. Dictionary Methods & Shallow Copy

### Common Methods:
- `.update()`: Updates the dictionary with elements from another dictionary or iterable of key-value pairs.
- `.clear()`: Removes all elements from the dictionary.

### The Concept of Shallow Copy (`.copy()`):
If you just use the assignment operator (`dict2 = dict1`), both variables point to the **same object in memory**. Changing one changes the other.
To create a separate copy, use the `.copy()` method. This creates a **shallow copy**. 

**Important note about Shallow Copies:** In a shallow copy, a new dictionary is created, but it inserts *references* to the objects found in the original. If your dictionary contains nested mutable objects (like lists or other dictionaries), modifying the nested object in the copy will ALSO affect the original.

```python
# update() method
d1 = {"a": 1, "b": 2}
d1.update({"b": 3, "c": 4}) # Updates 'b' and adds 'c'
print("After update:", d1)

# --- Understanding Shallow Copy ---
original = {
    "name": "Charlie", 
    "scores": [85, 90, 92] # This is a nested mutable object
}

# Creating a shallow copy
shallow = original.copy()

# Modifying a simple value in the copy (DOES NOT affect original)
shallow["name"] = "Charlie Updated"

# Modifying a nested mutable object in the copy (AFFECTS original!)
shallow["scores"].append(95)

print("\n--- Shallow Copy Results ---")
print("Original:", original)
print("Shallow Copy:", shallow)
# Notice how 'name' is different, but 'scores' got updated in BOTH!
```

## 6. Iterating Over Dictionaries
You can loop through a dictionary to get its keys, values, or both simultaneously.

```python
user = {"username": "admin", "role": "superuser", "active": True}

print("Iterating keys:")
for key in user:
    print(key)

print("\nIterating values:")
for value in user.values():
    print(value)

print("\nIterating both keys and values:")
for k, v in user.items():
    print(f"{k}: {v}")
```

## 7. Nested Dictionaries
A dictionary can contain other dictionaries. This is highly useful for structuring complex, hierarchical data like JSON files.

```python
family = {
  "child1" : {
    "name" : "Emil",
    "year" : 2004
  },
  "child2" : {
    "name" : "Tobias",
    "year" : 2007
  }
}

print("Family Dictionary:", family)
print("Accessing Child 2's name:", family["child2"]["name"])
```

## 8. Dictionary Comprehensions
Similar to list comprehensions, dictionary comprehensions provide a concise way to create dictionaries from iterables.

```python
# Creating a dictionary of squares for numbers 1-5
squares = {x: x*x for x in range(1, 6)}
print("Squares dictionary:", squares)

# Filtering with dictionary comprehension
even_squares = {x: x*x for x in range(1, 6) if x % 2 == 0}
print("Even squares only:", even_squares)
```

## 9. Practical Examples and Common Errors

### Practical Example: Frequency Counter
Dictionaries are perfect for counting occurrences of items (like words in a sentence).

```python
# Practical Example: Counting characters
text = "hello world"
char_count = {}
for char in text:
    if char != " ":
        # Use get() to provide a default of 0 if the key isn't there yet
        char_count[char] = char_count.get(char, 0) + 1

print("Character frequencies:", char_count)
```

### Common Errors to Avoid

1. **`KeyError`:** Occurs when you try to access a key that does not exist using square brackets `[]`.
   - **Solution:** Use the `.get()` method which safely returns `None` (or a specified default) instead of crashing.
2. **`TypeError` (Unhashable Type):** Occurs when you try to use a mutable data type (like a `list` or another `dict`) as a dictionary key. Keys MUST be immutable (like strings, integers, or tuples).
3. **Mistaking shallow copies for deep copies:** As shown earlier, `.copy()` doesn't fully protect nested data. If you need complete independence for nested lists/dicts, you must `import copy` and use `copy.deepcopy()`.

```python
# Demonstrating Errors (Wrapped in try-except to prevent the notebook from crashing)

# 1. KeyError
demo_dict = {"a": 1}
try:
    print(demo_dict["b"])
except KeyError as e:
    print(f"Caught a KeyError! Key {e} does not exist. Use .get() instead!")

# 2. TypeError (Unhashable Type)
try:
    # Trying to use a list as a key
    invalid_dict = {[1, 2]: "value"}
except TypeError as e:
    print(f"Caught a TypeError! {e}. You cannot use lists as dictionary keys.")
```
