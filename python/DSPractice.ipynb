```python
integer_list=[1,2,3,4,5]
```

```python
## Create a list containing 5 integers and print it.

print(integer_list)
```

```python
# Find the length of a list without using len().
count=0
for i in integer_list:
    count=count+1

print(count)
print("Length using len:",len(integer_list))
```

```python
# Print the first, last, and middle element of a list.

first=0
last=count-1
mid= first+((last-first)//2)
print(mid)
print("First:",integer_list[first])
print("Middle:",integer_list[mid])
print("Last:",integer_list[last])
```

```python
#Find the largest number in a list without using max().
maximum=integer_list[0]
for i in range(len(integer_list)):
    if(integer_list[i]>maximum):
        maximum=integer_list[i]

print(maximum)
max_value = max(integer_list)
print("Max using max():",max_value)
```

```python
#Find the smallest number without using min().
minimum=integer_list[0]
for i in range(len(integer_list)):
    if(integer_list[i]<minimum):
        minimum=integer_list[i]

print(minimum)
min_value = min(integer_list)
print("Min using max():",min_value)
```

```python
# Calculate the sum of all elements without using sum().
sum_value=0
for i in range(len(integer_list)):
    sum_value+=integer_list[i]

print(sum_value)
print("Sum using Sum",sum(integer_list))
```

```python
# Count how many even numbers are present in a list.
even_count=0
for i in range(len(integer_list)):
    if(integer_list[i]%2==0):
        even_count+=1

print(even_count)

#list comprehension
even_count_list=[i for i in range(len(integer_list)) if i%2==0]
print(even_count_list)
```

```python
# Count how many odd numbers are present.
odd_count=0
for i in range(len(integer_list)):
    if(integer_list[i]%2!=0):
        odd_count+=1

print(odd_count)

#list comprehension
odd_count_list=[i for i in range(len(integer_list)+1) if i%2!=0]
print(odd_count_list)
```

```python
#Reverse a list without using .reverse().
reverse_list=[]
list_length=len(integer_list)
while list_length >0:
    reverse_list.append(integer_list[list_length-1])
    list_length-=1

print(reverse_list)

# 1. Using Slicing
reverse_list_slicing = integer_list[::-1]
print("Reversed using slicing:", reverse_list_slicing)

# 2. Using reversed() function
reverse_list_func = list(reversed(integer_list))
print("Reversed using reversed():", reverse_list_func)

# 3. Using insert() in a loop
reverse_list_loop = []
for item in integer_list:
    reverse_list_loop.insert(0, item)
print("Reversed using insert():", reverse_list_loop)

'''insert(0, ...) is generally slow in Python for very large lists because Python has to 
rewrite the position of every single element in the memory every time you push something
 to the front. Slicing ([::-1]) is much faster and more efficient!'''
```
