# Python Conditionals

Conditionals help your program make decisions based on a true or false condition.

## Main ways to write conditionals in Python

### 1. `if`
Runs a block only when the condition is true.

### 2. `if-else`
Runs one block if the condition is true, otherwise runs another block.

### 3. `if-elif-else`
Checks multiple conditions one by one and runs the first true one.

### 4. Nested `if`
An `if` inside another `if` or `else` block. Use this when one decision depends on another.

### 5. Conditional expression
A short one-line form of `if-else`.
Example: `result = "Pass" if marks >= 40 else "Fail"`

### 6. `match-case` 7+
Useful for checking many fixed values, though it is different from normal `if-elif-else`.

## Common errors to avoid

- Using `=` instead of `==` in conditions
- Forgetting the colon `:` after `if`, `elif`, or `else`
- Wrong indentation
- Writing too many nested `if` statements when a simpler `if-elif-else` will do
- Not handling invalid user input
- Forgetting that `and` and `or` need complete conditions on both sides

## Example logic patterns

- Check age for eligibility
- Compare two numbers
- Validate user input
- Build menu-based programs

```python
print("Simple Calculator App")

num1 = float(input("Enter first number: "))
operator = input("Enter operator (+, -, *, /): ")
num2 = float(input("Enter second number: "))

if operator == "+":
    print("Result:", num1 + num2)
elif operator == "-":
    print("Result:", num1 - num2)
elif operator == "*":
    print("Result:", num1 * num2)
elif operator == "/":
    if num2 != 0:
        print("Result:", num1 / num2)
    else:
        print("Error: Division by zero is not allowed.")
else:
    print("Error: Invalid operator.")
```
