# Loops in Python

In Python, loops are used to iterate over a sequence (like a list, tuple, dictionary, set, or string) or execute a block of code multiple times. Python has two main types of loops:

1. **`for` loop**: Used for iterating over a sequence (like a list, string, or range).
2. **`while` loop**: Used for executing a block of code as long as a specified condition is true.

## The `range()` Function

The `range()` function returns a sequence of numbers, starting from 0 by default, increments by 1 (by default), and stops before a specified number.

**Syntax**: `range(start, stop, step)`
- `start` (optional): An integer specifying at which position to start. Default is 0.
- `stop` (required): An integer specifying at which position to stop (not included).
- `step` (optional): An integer specifying the incrementation. Default is 1.

```python
# Example of range()
for i in range(1, 6, 2):
    print(i) # Output: 1, 3, 5
```

## Iterating over a String

Strings are iterable objects, meaning you can loop through their characters one by one using a `for` loop.

```python
# Example of iterating a string
my_string = "Python"
for char in my_string:
    print(char)
```

## `break`, `continue`, and `pass`

### 1. `break`
Used to exit or "break out" of the loop entirely before it has looped through all the items.
**When to use**: When you have found what you are looking for or a specific condition is met, and you don't need to continue iterating.
```python
# Example of break
for i in range(5):
    if i == 3:
        break
    print(i) # Output: 0, 1, 2
```

### 2. `continue`
Used to skip the current iteration of the loop and continue with the next iteration.
**When to use**: When you want to skip executing the remaining code inside the loop for a specific condition but want to continue with the next loops.
```python
# Example of continue
for i in range(5):
    if i == 2:
        continue
    print(i) # Output: 0, 1, 3, 4
```

### 3. `pass`
Used as a placeholder when a statement is required syntactically but you do not want any command or code to execute.
**When to use**: When you are writing code and haven't implemented a loop body yet, or if you want an empty block of code.
```python
# Example of pass
for i in range(5):
    if i == 3:
        pass # To be implemented later
    print(i) # Output: 0, 1, 2, 3, 4
```
