## Comparison operators in Python

- == : equal to
- != : not equal to
-  ">"  : greater than
-  "<"  : less than
- ">=" : greater than or equal to
- <= : less than or equal to
- is / is not : identity comparison (same object)
- in / not in : membership check

Common errors to avoid:
- Using '=' instead of '==' (assignment vs comparison).
- Using 'is' to compare values (use '==' for value equality).
- Comparing incompatible types without intent (may be False or raise).
- Relying on identity (is) for small immutable caching/interning behavior.

```python
# comparison operators demo using existing var1 and var2
print("var1 =", var1, "var2 =", var2)
print("== :", var1 == var2)
print("!= :", var1 != var2)
print(">  :", var1 > var2)
print("<  :", var1 < var2)
print(">= :", var1 >= var2)
print("<= :", var1 <= var2)

# identity vs equality
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print("\na == b :", a == b)    # True (same contents)
print("a is b :", a is b)     # False (different objects)
print("a is c :", a is c)     # True (same object)

# membership
s = "hello"
lst = [10, 20, 30]
print("\n'e' in s     :", 'e' in s)
print("'x' not in s  :", 'x' not in s)
print("20 in lst     :", 20 in lst)
print("99 not in lst :", 99 not in lst)

# chained comparison
x = 5
print("\nChained 1 < x <= 10 :", 1 < x <= 10)
```
