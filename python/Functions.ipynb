# Python Functions

## 1. Introduction to Functions
A function is a block of organized, reusable code that is used to perform a single, related action. Functions provide better modularity for your application and a high degree of code reusing.

**Why use functions?**
- **Reusability:** Write once, use multiple times.
- **Organization:** Break down complex problems into smaller, manageable pieces.
- **Maintainability:** Easier to fix bugs or update logic in one place.

## 2. Defining Functions & 3. Calling Functions

- **Defining:** You define a function using the `def` keyword, followed by the function name, parentheses `()`, and a colon `:`.
- **Calling:** To execute the function, you "call" or "invoke" it by typing its name followed by parentheses.

```python
# Defining a simple function
def greet_user():
    """This is a docstring describing what the function does."""
    print("Hello! Welcome to the Python Functions notebook.")

# Calling the function
greet_user()
```

## 4. Function Parameters and Keyword Arguments
Information can be passed into functions as arguments (or parameters).

- **Positional Arguments:** Arguments that need to be included in the proper position or order.
- **Keyword Arguments:** Arguments passed by explicitly specifying the parameter name. This frees you from having to worry about the correct order.

```python
def describe_pet(animal_type, pet_name):
    print(f"\nI have a {animal_type}.")
    print(f"My {animal_type}'s name is {pet_name.title()}.")

# Positional Arguments (Order matters)
describe_pet('hamster', 'harry')

# Keyword Arguments (Order doesn't matter)
describe_pet(pet_name='willie', animal_type='dog')
```

## 5. Default Parameters
You can define a default value for a parameter. If the function is called without that argument, it uses the default value.

**Important Rule:** Parameters with default values must come *after* parameters without default values in the function definition.

```python
def describe_pet_default(pet_name, animal_type='dog'):
    print(f"\nI have a {animal_type}.")
    print(f"My {animal_type}'s name is {pet_name.title()}.")

# Using the default value for animal_type
describe_pet_default('willie')

# Overriding the default value
describe_pet_default('harry', 'hamster')
```

## 6. Variable-Length Arguments (`*args` and `**kwargs`)
Sometimes you don't know ahead of time how many arguments will be passed to your function.

- `*args` (Non-Keyword Arguments): Allows you to pass a variable number of positional arguments. They are gathered into a **tuple**.
- `**kwargs` (Keyword Arguments): Allows you to pass a variable number of keyword arguments. They are gathered into a **dictionary**.

```python
# Example of *args
def make_pizza(size, *toppings):
    print(f"\nMaking a {size}-inch pizza with the following toppings:")
    for topping in toppings:
        print(f"- {topping}")

make_pizza(12, 'pepperoni')
make_pizza(16, 'mushrooms', 'green peppers', 'extra cheese')

# Example of **kwargs
def build_profile(first, last, **user_info):
    """Build a dictionary containing everything we know about a user."""
    profile = {'first_name': first, 'last_name': last}
    for key, value in user_info.items():
        profile[key] = value
    return profile

user_profile = build_profile('albert', 'einstein', location='princeton', field='physics')
print("\nUser Profile:", user_profile)
```

## 7. Return Statement
The `return` statement is used to exit a function and go back to the place from where it was called. It can also pass data back to the caller.

- If there is no `return` statement, the function returns `None`.
- You can return multiple values by separating them with commas (they will be returned as a tuple).

```python
def get_formatted_name(first_name, last_name):
    full_name = f"{first_name} {last_name}"
    return full_name.title()

musician = get_formatted_name('jimi', 'hendrix')
print("Musician:", musician)

# Returning multiple values
def get_min_max(numbers):
    return min(numbers), max(numbers)

lowest, highest = get_min_max([10, 5, 20, 2, 8])
print(f"Lowest: {lowest}, Highest: {highest}")
```

## 8. Functions as Objects (Methods & Attributes)
In Python, functions are "first-class citizens". This means they are objects like any other data type (strings, lists, etc.). 
You can assign them to variables, pass them as arguments to other functions, and they have built-in attributes (methods).

```python
def shout(text):
    """This function converts text to uppercase."""
    return text.upper()

# 1. Assigning a function to a variable
yell = shout
print(yell("Functions are objects!"))

# 2. Accessing function attributes
print(f"Function Name: {shout.__name__}")
print(f"Function Docstring: {shout.__doc__}")
```

## 9. Practical Example
Let's combine these concepts into a practical example: A simple data processor.

```python
def process_student_data(name, *grades, **extra_info):
    """Processes a student's grades and extra information."""
    if not grades:
        average = 0
    else:
        average = sum(grades) / len(grades)
        
    report = {
        "Student Name": name,
        "Average Grade": round(average, 2),
        "Details": extra_info
    }
    return report

# Using positional, *args, and **kwargs all together
student_report = process_student_data("Alice", 85, 90, 92, 88, status="Honors", club="Robotics")
print("Student Report:", student_report)
```

## 10. Common Errors to Avoid

1. **`TypeError` (Missing Arguments):** Forgetting to pass required positional arguments.
2. **`SyntaxError` (Non-default argument follows default argument):** Placing a parameter without a default value *after* one that has a default value in the function definition.
3. **Scope Errors (`UnboundLocalError` or `NameError`):** Trying to access variables created inside a function from outside of it, or trying to modify a global variable inside a function without the `global` keyword.

```python
# Demonstrating Errors with try-except blocks

# 1. TypeError (Missing required positional arguments)
def multiply(a, b):
    return a * b

try:
    multiply(5) # Missing 'b'
except TypeError as e:
    print(f"Caught TypeError: {e}")

# 2. SyntaxError
# This cannot be caught in a try-except because it prevents the code from compiling.
# Incorrect: def register(age=18, name): 
# Correct:   def register(name, age=18):
print("\nReminder: Always put default parameters LAST in the function definition!")

# 3. Scope Error (NameError)
def calculate_tax(amount):
    tax = amount * 0.2
    return tax

calculate_tax(100)
try:
    # 'tax' is a local variable, it doesn't exist outside the function
    print(tax)
except NameError as e:
    print(f"\nCaught NameError: {e}. 'tax' is a local variable!")
```
